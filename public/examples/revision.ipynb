# 样本质量分析

这是用于演示审阅流程的合成数据。我们比较对照组与处理组的有效样本数，并核对缺失值处理。

```python
samples = [12, 15, None, 18, 21]
# 缺失项来自仪器未记录；本轮仅统计已记录样本。
valid = [value for value in samples if value is not None]
print(f"有效样本数：{len(valid)} / {len(samples)}")
```

## 结果说明

5 个合成样本中，4 个有完整记录，1 个因仪器未记录而缺失。本例仅展示审阅流程，不代表实验结论。

```python
mean = sum(valid) / len(valid)
print(f"平均值：{mean:.2f}")
```
