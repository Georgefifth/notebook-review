# Sample quality analysis

Synthetic data for demonstrating the review workflow. Check the number of recorded samples and how missing values are handled.

```python
samples = [12, 15, None, 18, 21]
# The instrument did not record the missing value; this review counts recorded samples only.
valid = [value for value in samples if value is not None]
print(f"Valid samples: {len(valid)} / {len(samples)}")
```

## Results

Of five synthetic samples, four have complete records and one is missing an instrument reading. This example demonstrates review mechanics, not experimental findings.

```python
mean = sum(valid) / len(valid)
print(f"Mean: {mean:.2f}")
```
